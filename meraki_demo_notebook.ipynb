# 🌿 Projet IA – Meraki Eco Intelligence Artificielle

Ce notebook montre comment utiliser une approche IA pour détecter les heures creuses dans un réseau Meraki à partir de données simulées.

```
# 📦 Import des bibliothèques
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
import random
```

```
# 🔁 Simulation de trafic réseau sur 24h
traffic_data = [random.randint(5, 100) for _ in range(24)]
plt.plot(range(24), traffic_data, marker='o')
plt.title('Trafic réseau simulé')
plt.xlabel('Heure')
plt.ylabel('Nombre de clients')
plt.grid(True)
plt.show()
```

```
# 🧠 Entraînement du modèle IA (Random Forest)
X = np.array([[i] for i in range(24)])
y = np.array(traffic_data)
model = RandomForestRegressor(n_estimators=10, random_state=42)
model.fit(X, y)
```

```
# ⚡ Prédiction et détection des heures creuses
predictions = model.predict(X)
low_traffic_hours = [i for i, val in enumerate(predictions) if val < 30]
print('Heures creuses détectées:', low_traffic_hours)
```

```
# 📊 Visualisation du résultat
plt.plot(range(24), predictions, marker='o', label='Prévision IA')
plt.axhline(y=30, color='red', linestyle='--', label='Seuil d’optimisation')
plt.title('Prévision du trafic horaire')
plt.xlabel('Heure')
plt.ylabel('Trafic estimé')
plt.legend()
plt.grid(True)
plt.show()
```

✅ Ce notebook constitue une preuve de concept pour prédire et visualiser les heures d’optimisation énergétique dans un réseau Cisco Meraki.
